```python
import pandas as pd
df_dep = pd.read_csv("/mnt/diversity/jiaqi/hmm/DenAdmix_Sim_git/downsampling/hmmix_ds.csv", sep = '\\s+')
frog_corr_dep = pd.read_csv("/mnt/diversity/jiaqi/hmm/DenAdmix_Sim_git/downsampling/frog_ds.csv", sep = '\\s+')
frog_corr_dep1 = frog_corr_dep[frog_corr_dep['dep']>=2]
df_dep1 = df_dep[df_dep['dep']>=2]
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")


samples = ['USR1',"Yana1","SS6004469","SS6004472"]
T = {
    "Yana1":"Yana1, Non-UDG, 5' c->t 15% ",
    "USR1": "USR1, UDG",
    "SS6004469":"B_Han-3",
    "SS6004472":"B_Papuan-15"
}
T = {
    "Yana1":"Yana1",
    "USR1": "USR1",
    "SS6004469":"B_Han-3",
    "SS6004472":"B_Papuan-15"
}


# Adjust figsize and layout as per the number of samples
fig, axes = plt.subplots(nrows=1, ncols=4, figsize=(16, 6), sharey=True)
axes = axes.flatten()

# Hide unused axes if fewer samples than subplots
for ax in axes[len(samples):]:
    ax.set_visible(False)

# Store handles and labels for the legend
handles, labels = [], []

for ax, sample in zip(axes, samples):
    df_dep_ = df_dep1[df_dep1['sample'] == sample]
    frog_corr_dep_ = frog_corr_dep1[frog_corr_dep1['sample'] == sample]
    line1, = ax.plot(df_dep_['dep'], df_dep_['cor'], label=f'{sample}',
            color='purple',
            marker = 'o',
            linestyle='-',
            markersize=4)
    handles.append(line1)
    labels.append(f'hmmix')
    
    line2, = ax.plot(frog_corr_dep_['dep'], frog_corr_dep_['cor'], label=f'{sample}',
            color='grey',
            marker = 's',
            linestyle='-',
            markersize=4)
    handles.append(line2)
    labels.append(f'admixfrog')
    ax.set_title(f'{T[sample]}', fontsize = 13)

tick_step = 5  # Adjust this step based on the data
fig.supylabel('Correlation coefficients of inferred Denisovan segments', fontsize = 14)
fig.supxlabel('Downsampled coverage', fontsize=14)

# Set x-ticks for all subplots

t = [[2,4,6,8,10,20],[2,4,6,8,10, 15, 30],[2,4,6,8,10,20,40],[2,4,6,8,10,20,40]]
for ax,t_t in zip(axes, t):
    ax.set_xticks(t_t)
    ax.set_xticklabels(t_t, fontsize = 12)
# Create a single shared legend
fig.legend(handles[0:2], labels[0:2], ncol = 1, loc='upper left',  fontsize = 12, bbox_to_anchor=(0.7, 0.1))

plt.tight_layout()
plt.subplots_adjust(top=0.85)  # Adjust the top to make room for the legend
plt.show()
# print(df[df['low'] ==3][['sample','ds',"anno_nea_corr4_FP1","anno_nea_corr4_FP2",'anno_corr4_FP1','anno_corr4_FP2','anno_nea4_split','anno_nea4_merge']])
# frog_corr
```
