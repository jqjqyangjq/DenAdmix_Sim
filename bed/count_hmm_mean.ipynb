```python
import pandas as pd
from collections import defaultdict
ind_geo = defaultdict(list)
import numpy as np
```

```python
OCE = ['S_Bougainville-1',
 'S_Bougainville-2',
 'S_Papuan-1',
 'S_Papuan-10',
 'S_Papuan-11',
 'S_Papuan-13',
 'S_Papuan-14',
 'S_Papuan-2',
 'S_Papuan-3',
 'S_Papuan-4',
 'S_Papuan-5',
 'S_Papuan-6',
 'S_Papuan-7',
 'S_Papuan-8',
 'S_Papuan-9',
 'B_Australian-3',
 'B_Australian-4',
 'B_Papuan-15']
from collections import defaultdict
ind_geo = defaultdict(str)
with open("Sample_all.sort.nonAFR.txt",'r') as f:
    for line in f:
        _, ind, pop, geo = line.strip().split()
        if geo == "Oceania":
            if ind in OCE:
                ind_geo[ind] = "Oceania"
        else:
            ind_geo[ind] = geo
geo_len_frog = defaultdict(lambda:defaultdict(int))
geo_len_hmm = defaultdict(lambda:defaultdict(int))
with open("SGDP.ind.list", 'r') as f:
    for line in f:
        file = line.strip()
        ind = file.split(".bed")[0]
        if ind_geo[ind] in ['WestEurasia','CentralAsiaSiberia','EastAsia','SouthAsia','Oceania','America']:
            with open("{}".format(file), 'r') as f:
                for line in f:
                    _, _, _, pos_len, _, _, map_len, m, *_ = line.strip().split()
                    if m == "frog":
                        if int(pos_len) >= 40000 and float(map_len) > 0.0499:
                            geo_len_frog[ind_geo[ind]][ind]+=int(pos_len)
                    elif m == "hmm":
                        if int(pos_len) >= 40000 and float(map_len) > 0.0499:
                            geo_len_hmm[ind_geo[ind]][ind]+=int(pos_len)
                        
    
total_n = defaultdict(int)
for i,z in zip(['WestEurasia','CentralAsiaSiberia','EastAsia','SouthAsia','Oceania','America'],
               [75,26,47,39,18,21]):
    total_n[i] = z
```

```python
print("Frog")
for geo in list(geo_len_frog.keys()):
    t = 0
    lengths = np.zeros(len(geo_len_frog[geo]))
    for i, ind in enumerate(geo_len_frog[geo]):
        lengths[i] = geo_len_frog[geo][ind]
        t += geo_len_frog[geo][ind]
    print(geo, np.sum(lengths)/total_n[geo])
    
print("HMM")
for geo in list(geo_len_hmm.keys()):
    t = 0
    lengths = np.zeros(len(geo_len_hmm[geo]))
    for i, ind in enumerate(geo_len_hmm[geo]):
        lengths[i] = geo_len_hmm[geo][ind]
        t += geo_len_hmm[geo][ind]
    print(geo, np.sum(lengths)/total_n[geo])
    
```

```python
df_frog_count = pd.DataFrame(columns=['ind','total_len_filter_40000','number_seg_filter_40000'])
with open("check1", 'r') as f:
    for line in f:
        line = line.strip()
        with open(f"{line}.bed", 'r') as f:
            length = 0
            count = 0
            for record in f:
                _, _, _, pos_len, _, _, map_len, m, *_ = record.strip().split()
                if m == "frog":
                    if int(pos_len) >= 40000 and float(map_len) > 0.0499:
                        length += int(pos_len)
                        count += 1
            df_frog_count = pd.concat([df_frog_count, pd.DataFrame({'ind': [line.strip()], 'total_len_filter_40000': [length], 'number_seg_filter_40000': [count]})], ignore_index=True)
df_frog_count.to_excel("check1.frog.count.xlsx", index=False)
```
